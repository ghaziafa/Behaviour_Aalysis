 "Text redibility score" function measures how easy your text is to read by calculating:
Number of words,Number of sentences,Number of syllables,Then applying the standard readability formula.

```python
def read_sco(text):
    words = text.split()
    if not word:
        return 0.0
    sentences = max(1, text.count('.')+text.count('!')+text.count('?'))
    #count vowels per word
    syllable_count = sum(sum(1 for ch in word if ch in 'aeiou') for word in words)
    ## avoid division by zero
    words_per_sent = len(words)/sentences
    syllables_per_word = syllables_count/len(words) if len(words) else 0
    score = 206.835 - 1.015 * words_per_sentence - 84.6 * syllables_per_word
    return round(score, 2)
```

Report Generator


```python
def generate_report(analysis):
    report = []
    report.append("Psychological Analysis Report")
    report.append("-----------------------------")
    report.append(f"Sentiment: {analysis.get('sentiment')} (polarity: {analysis.get('sentiment_polarity')})")
    report.append("")
    report.append("Emotion Counts:")
    for emo, count in analysis.get('emotion_count', {}).items():
        report.append(f" - {emo}: {count}")
        report.append(f"Dominant Emotion: {', '.join(analysis.get('dominant_emotion', ['neutral']))}")
    report.append("")
    report.append("Cognitive Distortions Detected:")
    if analysis.get('cognitive_distortions'):
        for d in analysis.get('cognitive_distortions'):
            report.append(f" - {d}")
    else:
        report.append(" - None detected")
    report.append("")
    report.append("Linguistic Markers:")
    for m, val in analysis.get('linguistic_markers', {}).items():
        report.append(f" - {m}: {val}")
    report.append("")
    report.append(f"Readability Score: {analysis.get('readability_score')}")
    report.append("-----------------------------")
    return "\n".join(report)
```
